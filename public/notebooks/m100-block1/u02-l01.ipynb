## Python powers and safe roots

Python writes powers with **. Exponentiation binds more tightly than unary minus, so -3**2 gives −9; (-3)**2 gives 9. An import makes a library available: import math lets you call math.sqrt(...). The dot selects a function provided by that library; you are using a supplied function, not yet defining your own. math.sqrt expects a non-negative real input. A ValueError indicates an unacceptable value for that operation. Do not treat a fractional power of a negative base in Python as a general real odd-root calculator.
