## Understand the relationship

A reproducible notebook contains the question, data/source, assumptions, code, labelled figures, checks and limitations. Restart and run it in order. An agent is software that can propose or execute multi-step actions; a suggestion is not evidence of correctness. In a disposable copy, permit one bounded change such as adding axis labels, inspect the diff, run independent tests and reject or roll back any change to data or model. For this free prototype, use the supplied simulated patch rather than installing or paying for an agent. This rehearses review, not proficiency with a live agent.
