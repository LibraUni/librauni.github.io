## Boolean comparisons and if/else

Python Boolean values are True and False. Comparisons use == for equality, != for inequality, and <, >, <=, >=. A single = assigns a value; it does not compare. An if line ends with a colon. The indented statements run only when its condition is true. An else line, also ending with a colon, selects the other branch. Use four spaces consistently. The protected calculation must be inside the valid branch, not before it.
