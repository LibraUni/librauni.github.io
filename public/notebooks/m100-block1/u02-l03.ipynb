## A conversion that does not hide its units

Ordinary Python arithmetic does not know that a number represents kilometres, hours or metres. You must perform conversions explicitly and document the convention. A correctly executed calculation with mismatched units can be physically wrong. Put unit conversion before combining quantities, and compare with a hand-calculated result.
