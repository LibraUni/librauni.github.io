## A notebook that reruns cleanly

A reproducible notebook records the complete route from inputs to interpretation. Put assumptions and predictions in text cells, then type the code. Use meaningful variable names and print quantities with labels. To check reproducibility, save, restart the kernel and run cells from top to bottom. If a later cell depends on an earlier definition, the earlier cell must be part of the notebook. Saving a file does not itself make an off-device backup.
