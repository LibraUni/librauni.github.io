## Names, assignment and printed values

A Python name is a label attached to a value. In elapsed = 12, Python evaluates the right-hand side and assigns its value to the name on the left. This is an instruction, not an equation to solve. Later elapsed = 15 replaces the value associated with that name. Integers such as 12 have type int; decimal literals such as 12.0 have type float. Ordinary / division produces a float, even when the mathematical quotient is an integer. print(...) explicitly displays a value; a notebook also normally displays the final expression in a cell. Choose names that convey meaning and use an underscore for spaces. A notebook remembers previous execution, so run cells in order when checking your work.
