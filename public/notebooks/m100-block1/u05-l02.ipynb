## Understand the relationship

NumPy arrays store numerical sequences and apply arithmetic element by element. import introduces a library; as gives it a short name. Matplotlib plots corresponding x/y values; labels and units are part of the result. np.loadtxt reads a simple numeric CSV: delimiter names the separator and skiprows=1 skips its header. A[:, 0] selects all rows of the first column; indexing begins at zero. Use the supplied CSV beside your notebook.
