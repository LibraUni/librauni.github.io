## Understand the relationship

A def statement defines a function; the indented body runs when it is called. Arguments supply inputs; return sends a result back to the caller. print only displays something. A local name belongs to that call. A docstring is a quoted description at the start. Test ordinary, zero and negative inputs with expected results chosen independently. A test is evidence about those cases, not a proof for every real input.
