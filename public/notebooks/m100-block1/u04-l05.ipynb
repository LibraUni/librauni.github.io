## Lists and repeated candidate checks

A Python list stores an ordered sequence of values inside square brackets, separated by commas. A for loop visits them in order. In “for x in candidates:”, x takes each listed value in turn; the indented body runs once for each. The colon and indentation matter as in if. Start with explicit candidates, especially boundaries. Finite checks illustrate an inequality but cannot establish every real solution.
