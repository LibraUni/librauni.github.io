## Scientific notation in code

Python’s e notation means “times ten to the power”: 3e4 is 30000.0, not 3 raised to 4. It creates a floating-point value. Use names to keep units visible, such as distance_m. Those names are reminders for humans, not an automatic unit system. Retain the computed value while deciding separately how to report it. A decimal display in e notation has not changed the underlying physical quantity.
