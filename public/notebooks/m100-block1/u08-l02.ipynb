## Understand the relationship

In y=A sin(ωt+φ)+c, |A| is amplitude, c the centre, and the period is 2π/|ω| when ω≠0. Python/NumPy trig uses radians. An inverse trig function returns a principal value; further angles can solve the same equation on a stated interval. For sinθ=1/2 on [0,2π), the solutions are π/6 and5π/6.
