## Assertions that check a claim

An assert statement checks a condition and stops with AssertionError if it is false. If true, it produces no output. Use it for a specific expected property, with an explanatory message after a comma. Here the values are small integers, so exact equality is appropriate. Do not treat exact float equality as a general numerical accuracy test. An assertion checks what you wrote; a mistaken expected value remains a mistaken check.
