## Understand the relationship

The dot product a·b=aₓbₓ+aᵧbᵧ also equals |a||b|cosθ. It is a scalar. The signed projection of a onto a unit direction u is a·u; the vector projection is (a·u)u. NumPy arrays support component operations, np.dot computes the dot product and np.linalg.norm the magnitude. Check numerical results against a simple diagram and an exact calculation.
