### Load and plot — 20 minutes
The three blocks below match the worked lesson. Read its explanation of every new command. A successful load does not establish correct interpretation.

```python
import numpy as np

data = np.loadtxt(
    "p101-u03-l02.csv",
    delimiter=",",
    skiprows=1,
    ndmin=2,
)
print("Array shape (rows, columns):", data.shape)
print(data)
```

```python
time_s = data[:, 0]
position_m = data[:, 1]
print("First position / m:", data[0, 1])
print("C clock reading / s:", data[2, 0])
elapsed_s = time_s - time_s[0]
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
ax.plot(time_s, position_m, marker="o", linestyle="none")
ax.set_xlabel("Clock reading / s")
ax.set_ylabel("Position / m")
ax.set_title("Invented carriage record P101-U03-L01-A")
ax.grid(alpha=0.3)
fig.savefig("p101-u03-l02-clock.png", dpi=160)
plt.show()
```
